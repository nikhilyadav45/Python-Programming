```python
# This is the simple bank system
balance = 1000
account_number = 3754
u_name = ""

def create_acc():
    print("You Selected Create Account ->")
    u_name = str(input("Enter your name:"))
    print(f"Thank you for creating account {u_name}")
    return 100
def deposit_mo(balance):
    deposit_amount = int(input("Enter the amount to Deposit"))
    balance = balance + deposit_amount
    print(f"Your total Balance is : {balance}")
def withdraw_mo(balance):
    cass = int(input("Enter the amount to Deposit"))
    if cass < balance:
        balance = balance - cass
        print(f"Your total Balance is : {balance}\n Please collect your cass")
    else:
        print("Insuficient balance")
def check_ball(balance):
    print(f"Your total Balance is : {balance}")
def exit():
    print("Thank you for using our banking system")

print("\n ==============================================================")
print("                           BANKING SYSTEM                      ")
print(" ==============================================================\n")
account = int(input("Enter the 4 digit Account no."))
if account == account_number:
    print("1. Create an account")
    print("2. Deposit money")
    print("3. Withdraw money")
    print("4. Check balance")
    print("5. Exit")
    print(" ==============================================================\n")
    task = int(input("Enter the task number: "))
    if task == 1:
        balance = create_acc()
        print(f"Your Bank account is created with balance: {balance}")
    elif task == 2:
        deposit_mo(balance)
    elif task == 3:
        withdraw_mo(balance)
    elif task == 4:
        check_ball(balance)
    elif task == 5:
        exit()
    else:
        print("Invalid task number")
else:
    print("Invalid Account number")
```
