```python
# Initialize the primary database using a Dictionary
# Key: Roll Number (Integer) -> Value: Another Dictionary containing profile data
student_database = {}

def add_student(roll_no, name, dob, courses_set, marks_list):
    """
    Adds a new student to the database.
    Uses a Tuple for personal details, a Set for courses, and a List for marks.
    """
    if roll_no in student_database:
        print(f"Error: Student with Roll Number {roll_no} already exists.")
        return

    # Tuple holds fixed personal data (Immutable)
    personal_details = (name, dob)
    
    # Set ensures unique course entries (No duplicates)
    enrolled_courses = set(courses_set)
    
    # List tracks sequential grades over time (Mutable)
    academic_marks = list(marks_list)

    # Store everything inside a nested profile dictionary
    student_database[roll_no] = {
        "personal": personal_details,
        "courses": enrolled_courses,
        "marks": academic_marks
    }
    print(f"Success: Student {name} added successfully.")

def update_marks(roll_no, new_mark):
    """Updates the mutable marks List by appending a new grade."""
    if roll_no in student_database:
        student_database[roll_no]["marks"].append(new_mark)
        print(f"Success: Added mark {new_mark} to Roll Number {roll_no}.")
    else:
        print("Error: Student record not found.")

def register_course(roll_no, course_code):
    """Adds a course to the student's Course Set."""
    if roll_no in student_database:
        # Sets automatically handle duplicate elimination
        student_database[roll_no]["courses"].add(course_code)
        print(f"Success: Registered for {course_code}.")
    else:
        print("Error: Student record not found.")

def display_database():
    """Iterates through the Dictionary and prints the student rosters cleanly."""
    if not student_database:
        print("The database is currently empty.")
        return

    print("\n" + "="*50)
    print("                STUDENT DATABASE")
    print("="*50)
    
    # Iterating over the main storage Dictionary
    for roll_no, data in student_database.items():
        # Unpacking the immutable personal Tuple
        name, dob = data["personal"]
        
        # Accessing the Course Set and Marks List
        courses = ", ".join(data["courses"])
        marks = data["marks"]
        avg_mark = sum(marks) / len(marks) if marks else 0.0

        print(f"Roll Number : {roll_no}")
        print(f"Name        : {name}")
        print(f"DOB         : {dob}")
        print(f"Courses     : {courses if courses else 'None'}")
        print(f"Marks History: {marks}")
        print(f"Average GPA : {avg_mark:.2f}")
        print("-" * 50)

# ==========================================
# Execution and Testing
# ==========================================

# 1. Populating the database with sample data
add_student(
    roll_no=101, 
    name="Rahul Sharma", 
    dob="2004-05-12", 
    courses_set={"CS101", "MA102", "CS101"}, # "CS101" duplicated on purpose
    marks_list=[85, 90, 78]
)

add_student(
    roll_no=102, 
    name="Priya Patel", 
    dob="2003-11-23", 
    courses_set={"EE101", "MA102"}, 
    marks_list=[92, 88]
)

# 2. Modifying the database attributes dynamically
print("\n--- Performing Live Database Modifications ---")
register_course(101, "PH101")   # Adding a brand new course to the Set
register_course(101, "CS101")   # Attempting a duplicate course insertion
update_marks(102, 95)           # Appending a new exam score to the List

# 3. Viewing the final optimized dataset
display_database()
```
