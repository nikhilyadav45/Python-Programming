Make Programe to calculate the two number just like calculater.

```python
num1 = int(input("Please enter the first number"))
num2 = int(input("Please enter the second number"))
operation = input("Please enter the operation symbol")
result = 0

if operation == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operation == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")
elif operation == "/":
    result = num1 / num2
    print(f"{num1} / {num2} = {result}")
else:
    print("Please enter the valid operation")
```

Number Guessing Game

```python
import random
print("Number guessing game between 1 to 100")
attempts = 5
number = random.randint(1,100)
print(f"win number {number}")

while(attempts > 0):
    print(f"Your attempt number is {attempts}")
    guess = int(input("Enter the number: "))
    if guess > number:
        print("Too high! Try again")
    elif guess < number:
        print("Too low! Try again")
    else:
        print("You Win!")
        break
    
    attempts = attempts - 1
    if(attempts == 0):
        print("Better Luck next time......")


```
